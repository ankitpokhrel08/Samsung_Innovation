```python
#Q. 33-01
words = ["A", "MERRY", "XMAS", "TO", "ALL"]
unique_letters = sorted(set("".join(words)))  # Unique letters in the words

def is_perfect_square(number):
    x = 0
    while x * x <= number:
        if x * x == number:
            return True
        x += 1
    return False

def convert_word_to_number(word, letter_to_digit):
    number = 0
    for char in word:
        number = number * 10 + letter_to_digit[char]
    return number

def is_mapping_valid(letter_to_digit):
    for word in words:
        word_number = convert_word_to_number(word, letter_to_digit)
        digit_sum = sum(int(digit) for digit in str(word_number))
        if not is_perfect_square(word_number) or not is_perfect_square(digit_sum):
            return False
    return True

fixed_mappings = {'T': 8, 'O': 1}
remaining_letters = [letter for letter in unique_letters if letter not in fixed_mappings]

solution_found = False

def find_solution(current_index, letter_to_digit, used_digits):
    global solution_found
    if current_index == len(remaining_letters):
        if is_mapping_valid(letter_to_digit):
            print("Valid Letter-to-Digit Mapping Found:")
            for letter, digit in letter_to_digit.items():
                print(f"{letter} -> {digit}")
            solution_found = True
        return

    current_letter = remaining_letters[current_index]

    for digit in range(10):
        if digit not in used_digits:
            letter_to_digit[current_letter] = digit
            used_digits.add(digit)
            find_solution(current_index + 1, letter_to_digit, used_digits)
            if solution_found:
                return
            del letter_to_digit[current_letter]
            used_digits.remove(digit)

find_solution(0, fixed_mappings.copy(), set(fixed_mappings.values()))

if not solution_found:
    print("No valid mapping found.")
```
